```python
import random, timeit, numpy as np
n = 10_000_000
a = [random.random() for _ in range(n)]; b = [random.random() for _ in range(n)]
A, B = np.array(a), np.array(b)
loop = min(timeit.repeat(lambda: sum(x * y for x, y in zip(a, b)), number=1, repeat=3))
vec = min(timeit.repeat(lambda: A @ B, number=1, repeat=20))
print(f"loop {loop:.3f}s  numpy {vec*1e3:.2f}ms  ratio {loop/vec:.0f}x")
```

## Reproduce the numbers quoted in the text

```python
from numba import njit

@njit
def dot(x, y):
    s = 0.0
    for i in range(x.size):
        s += x[i] * y[i]
    return s

dot(A, B)                                   # compile once
jit = min(timeit.repeat(lambda: dot(A, B), number=1, repeat=20))
print(f"numba {jit*1e3:.1f}ms: {loop/jit:.0f}x faster than the loop, numpy {jit/vec:.0f}x faster than numba")
col, row = np.zeros((4, 1)), np.zeros(4)
print("(B, 1) + (B,) broadcasts to", (col + row).shape)
```
